### 1.1 Find the average insurance charges of the based on soutwest region

```python
employee_data["region"].unique()
```

```python
employee_data.groupby("region")["charges"].mean().round(2).sort_values(ascending = False)
```

### 1.2 Find the average insurance charges of the based on region and age

```python
## groupby(["region","sex"]) - Discrete values (should be within the groupby function)    
##                         ["charges"] - Continous values, it should be measurable and have units (has to be outside the group by function in [])

employee_data.groupby(["region","sex"])["charges"].mean().round(2).sort_values(ascending = False)
```

### 2. Pivot Table

```python
### Used when need a presenting 2 or more categorical values{in indexx and columns} with a discrate value {in values}

pd.pivot_table(data = employee_data, values = "charges", index = "region", columns = "sex").round()
```

## 3. Aggregate Methods

### 3.1 Find the avg, min,max, sum of insurance charges of the based on region and sex

```python
employee_data.head()
```

```python
employee_data.groupby(["region", "sex"])["charges"].agg(func = ["mean", "sum", "min", "max"]).round()
```

### 4. CrossTab

To compute the frequency of two categorical features(discrete values like age, sex, region something which can not be measured) we use crosstab.

## 4.1 How many smokers and non smokers based on regions?

```python
pd.crosstab(index = employee_data["region"], columns = employee_data["smoker"], margins = True)
```

## 5. Apply Function

```python
def get_age(x):
    if x > 20:
        return "Eligible"
    else:
        return "Not eligible"
```

```python
employee_data["Eligibility Criteria"] = employee_data["age"].apply(get_age)
employee_data
```

## 6. Lamba Function

Nameless or annonamus Function

```python
employee_data["EC"] = employee_data["age"].apply(lambda x:1 if x>20 else 0)
employee_data
```
